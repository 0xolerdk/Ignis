# Segmentation Ablation Study (Synthetic)

Simple comparison of dropout and horizon configurations using random metrics to validate reporting utilities.

```python
import pandas as pd
import numpy as np
rng = np.random.default_rng(123)
configs = [
    {"name": "baseline", "dropout": 0.1, "dice": 0.72, "brier": 0.18},
    {"name": "dropout_0.3", "dropout": 0.3, "dice": 0.75, "brier": 0.16},
    {"name": "mc128", "dropout": 0.3, "dice": 0.76, "brier": 0.15},
]
df = pd.DataFrame(configs)
df
```

## Relative improvement

```python
baseline = df.loc[0]
df.assign(
    dice_gain=lambda d: d.dice - baseline.dice,
    brier_gain=lambda d: baseline.brier - d.brier,
)
```

## Recommendation

```python
best = df.sort_values('dice', ascending=False).iloc[0]
print(f"Recommended configuration: {best.name} (dropout={best.dropout})")
```
